# NUMPY and MATRICES

```python
import numpy as np
```

### BASICS


```python
one_dim_arr = np.array([12,20])
print(one_dim_arr)

# creating array from 3 ints starting from 0
arr = np.arange(3)
print(arr)

#starts with one, ends at 20, incremented by 3
arr = np.arange(1,20,3)
print(arr)

#Arrays with evenly spaced values from a given range of numbers: linspace (linear space)
arr = np.linspace(0,100,5, dtype=int) # start, end, number of elements. specified dtype as  float64 is default
print(arr)
print(type(arr[0]))
```

### Multi-dimensional Arrays

```python
one_dim_arr = [6,5,4,3,2,1]
two_dim_arr = np.array([[1,2,3],[4,5,6]])
print(two_dim_arr)

#reshaping 1d array to multi dim array
arr = np.reshape(
    one_dim_arr,
    (2,3)
)
print(arr)
```

```python
print(arr.shape) # m x n
print(arr.size)  # number of elements
print(arr.ndim)  # dimensions
```

### Array Math Operations

```python
arr_1 = np.array([2, 4, 6])
arr_2 = np.array([1, 3, 5])
```

```python
#adding subtracting 2 1d arrays
sum = arr_1+arr_2
diff = arr_1 - arr_2
print(sum)
print(diff)

# multiplication
prod = arr_1*arr_2
print(prod)
```

```python
# Scalar vector multiplicaton (Broadcasting)
print(arr_1 * 3)
```

### Stacking

```python
a1 = np.array([[1,1], 
               [2,2]])
a2 = np.array([[3,3],
              [4,4]])
print(f'a1:\n{a1}')
print(f'a2:\n{a2}')
```

```python
vert_stack = np.vstack((a1,a2))
print(vert_stack)
```

```python
hor_stack = np.hstack((a1,a2))
print(hor_stack)
```
